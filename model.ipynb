```python
data = [
    {"rank": 1, "model": "Mahalanobis Distance",              "type": "Anomaly detector", "labels": "None", "precision": 0.938, "recall": 0.847, "f1": 0.890},
    {"rank": 2, "model": "Autoencoder",                       "type": "Anomaly detector", "labels": "None", "precision": 0.928, "recall": 0.842, "f1": 0.883},
    {"rank": 3, "model": "Isolation Forest (tuned)",          "type": "Anomaly detector", "labels": "None", "precision": 0.932, "recall": 0.705, "f1": 0.802},
    {"rank": 4, "model": "Isolation Forest",                  "type": "Anomaly detector", "labels": "None", "precision": 0.931, "recall": 0.677, "f1": 0.784},
    {"rank": 5, "model": "Logistic Regression, baseline",     "type": "Supervised",       "labels": "5%",   "precision": 0.917, "recall": 0.618, "f1": 0.739},
    {"rank": 6, "model": "Logistic Regression, self-trained", "type": "Semi-supervised",  "labels": "5%",   "precision": 0.916, "recall": 0.615, "f1": 0.736},
    {"rank": 7, "model": "Logistic Regression, baseline",     "type": "Supervised",       "labels": "1%",   "precision": 0.915, "recall": 0.606, "f1": 0.729},
    {"rank": 8, "model": "Logistic Regression, self-trained", "type": "Semi-supervised",  "labels": "1%",   "precision": 0.912, "recall": 0.601, "f1": 0.720},
]

df = pd.DataFrame(data)
```

```python
df.head(8)
```

```python
df["label"] = df.apply(
    lambda r: r["model"] if r["labels"] == "None" else f"{r['model']} ({r['labels']})",
    axis=1,
)

long_df = df.melt(
    id_vars=["rank", "label", "model", "type", "labels"],
    value_vars=["precision", "recall", "f1"],
    var_name="metric",
    value_name="score",
)

fig = px.bar(
    long_df,
    x="label",
    y="score",
    color="metric",
    barmode="group",
    hover_data=["rank", "model", "type", "labels"],
    text="score",
    category_orders={"label": df.sort_values("rank")["label"].tolist()},
    title="Model Performance Comparison",
)
fig.update_traces(texttemplate="%{text:.3f}", textposition="outside")
fig.update_layout(yaxis_range=[0.5, 1.0], xaxis_title="", yaxis_title="Score", xaxis_tickangle=-30)
fig.show()
```

```python
fig.write_html("model_performance.html", include_plotlyjs=True)
```
